```python
def calculate_discount(price, discount_percent=10):
    discount = ((discount_percent/ 100)* price)
    final_price = price - discount
    return final_price

print(f"new price is {calculate_discount(800,discount_percent=20)}")
#print("new price is", calculate_discount(800, discount_percent))
print(f"new price is {calculate_discount(800,discount_percent=25)}")
print(f"new price is {calculate_discount(800,discount_percent=0)}")
```

```python
import random

# Your code here:
def add_result():
    num_1 = random.randint(1, 6)
    print(num_1)
    num_2 = random.randint(1,6)
    print(num_2)

    return num_1 + num_2
ans = add_result()
print(ans)
student_names = ["Ade", "Fola", "Mimi", "John", "Sky"]
print(random.choice(student_names))


def get_random_n(start, digit_count):
    num_4digits = ""
    for i in range(start, digit_count):
         num = f"num_(i)"
         num = random.randint(1, 10)
         num_4digits += str(num)
    return(num_4digits)
otp = get_random_n(1, 5)
print(otp)
```

```python
def get_random_n(start, digit_count, max_digit):
    num_4digits = ""
    for i in range(start, digit_count):
         num = f"num_(i)"
         num = random.randint(start, max_digit)
         num_4digits += str(num)
    return(num_4digits)
otp = get_random_n(1, 5, 9)
print(otp)
```
