```python
def calculate_discount(price, discount_percent=10):
    discount = ((discount_percent/ 100)* price)
    final_price = price - discount
    return final_price

print(f"new price is {calculate_discount(800,discount_percent=20)}")
#print("new price is", calculate_discount(800, discount_percent))
print(f"new price is {calculate_discount(800,discount_percent=25)}")
print(f"new price is {calculate_discount(800,discount_percent=0)}")
```
